```python
import os
import shutil

# --- CẤU HÌNH ---
# 1. Đường dẫn thư mục code đã clone (Mặc định thường là /content/VM-UNet)
repo_path = '/content/VM-UNet'

# 2. Đường dẫn trên Drive của bạn
drive_path = '/content/drive/MyDrive/VM-UNet' # Sửa lại nếu bạn để ở folder khác
zip_name = 'isic17_256.zip'
weight_name = 'vmamba_small_e238_ema.pth'

# --- BẮT ĐẦU XỬ LÝ ---
print(f"📂 Đang làm việc tại: {repo_path}")

# 1. Copy & Setup Pre-trained Weights
# Config tìm file tại: ./pre_trained_weights/vmamba_small_e238_ema.pth
print("⬇️  Đang thiết lập pre-trained weights...")
weight_dest_dir = os.path.join(repo_path, 'pre_trained_weights')
os.makedirs(weight_dest_dir, exist_ok=True)

try:
    shutil.copy(os.path.join(drive_path, weight_name),
                os.path.join(weight_dest_dir, weight_name))
    print("✅ Copy Weights thành công.")
except FileNotFoundError:
    print(f"❌ LỖI: Không tìm thấy file {weight_name} trong Drive tại {drive_path}")

# 2. Copy & Giải nén Dataset
# Config tìm data tại: ./data/isic17/
print("⬇️  Đang copy dataset từ Drive...")
zip_src = os.path.join(drive_path, zip_name)
zip_dst = os.path.join(repo_path, zip_name) # Copy thẳng vào thư mục dự án để giải nén

if os.path.exists(zip_src):
    shutil.copy(zip_src, zip_dst)

    print("📦 Đang giải nén dataset...")
    # Vì file zip của bạn đã chứa folder 'data/isic17', ta giải nén thẳng vào root của repo
    # Lệnh unzip này sẽ merge folder 'data' trong zip vào folder 'data' của repo (nếu có)
    os.system(f'unzip -q {zip_dst} -d {repo_path}')

    # Xóa file zip cho nhẹ máy
    os.remove(zip_dst)

    # Kiểm tra lại kết quả
    expected_path = os.path.join(repo_path, 'data', 'isic17')
    if os.path.exists(os.path.join(expected_path, 'train')) and os.path.exists(os.path.join(expected_path, 'val')):
        print(f"✅ Giải nén thành công! Data nằm tại: {expected_path}")
    else:
        print(f"⚠️ Cảnh báo: Cấu trúc giải nén có vẻ chưa đúng. Hãy kiểm tra folder {os.path.join(repo_path, 'data')}")
else:
    print(f"❌ LỖI: Không tìm thấy file {zip_name} trong Drive.")

print("\n------------------------------------------------")
print("Sẵn sàng để train! Hãy cd vào thư mục dự án và chạy lệnh train:")
print(f"cd {repo_path}")
print("python train.py")
```
